```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler
```

```python
def dj_circuit(n, oracle):

    #n inuput qubits, 1 ancilla, n classical bit
    qc= QuantumCircuit(n+1, n)
    # start the ancilla in |1>
    qc.x(n)
    # apply hadamard to everything (ancilla become |->)
    qc.h(range(n+1))
    qc.barrier()
    # the single oracle call
    oracle(qc,n)
    qc.barrier()
    #h to inputs only
    qc.h(range(n))
    #measure inputs only
    qc.measure(range(n), range(n))
    return qc
```

```python
# f(x) = x0 XOR x1 XOR x2

def balanced(qc, n):
    for i in range(n):
        qc.cx(i,n)
```

```python
#f(x) = 1 for every x
def constant_one(qc, n):
  qc.x(n)

sampler = StatevectorSampler()
for name, oracle in [("balanced", balanced), ("constant", constant_one)]:
  qc=dj_circuit(3, oracle)
  counts = sampler.run([qc], shots=1024).result()[0].data.c.get_counts()
  print(name, counts)
```
